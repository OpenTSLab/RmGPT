```python
import pickle
import os
all_result = pickle.load(open('classification_result.pkl', 'rb'))
```

```python
from collections import Counter
import matplotlib.pyplot as plt
file_num = 0
confident_num =0
for file, results in all_result.items():
    counter = Counter(results)
    total = len(results)
    # print(f"For file {file}:")
    # if len(counter.items())>=2:
    #     plt.figure()
    #     plt.title(file)
    #     plt.plot(range(len(results)), results)
        
    for number, count in counter.items():
        percentage = 100 * count / total
        print(f"Number {number} appears {count} times, which is {percentage:.2f}% of the total.")
        # if percentage >= 90:
        #     confident_num += 1
```

```python
#找到文件里数目最多的分类
import pandas as pd
result=pd.DataFrame(columns=['Sample Number','Fault Code'])
for file,results in all_result.items():
    counter = Counter(results)
    total = len(results)
    max_num = max(counter, key=counter.get)
    #把file中的Sample字段去掉
    result=result.append({'Sample Number':int(file.strip('Sample')),'Fault Code':max_num},ignore_index=True)
```

```python
label_code = {
            'TYPE0': 0,
            'TYPE1': 1,
            'TYPE2': 2,
            'TYPE3': 3,
            'TYPE4': 4,
            'TYPE5': 5,
            'TYPE6': 6,
            'TYPE7': 7,
            'TYPE8': 8,
            'TYPE9': 9,
            'TYPE10': 10,
            'TYPE11': 11,
            'TYPE12': 12,
            'TYPE13': 13,
            'TYPE14': 14,
            'TYPE15': 15,
            'TYPE16': 16,
        }
label_code_reversed = {v: k for k, v in label_code.items()}

result.sort_values(by='Sample Number')
#重新组织index
result.reset_index(drop=True,inplace=True)
result['Fault Code']=result['Fault Code'].map(label_code_reversed)
```

```python
result.to_csv('result.csv')
```
